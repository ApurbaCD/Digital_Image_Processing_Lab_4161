```python
import numpy as np
import matplotlib.pyplot as plt

# Input pixel values normalized between 0 and 1
r = np.linspace(0, 1, 256)

# Logarithmic transformation
c_log = 2
s_log = c_log * np.log(r+1)  # log(r+1)

# Power law (gamma) transformations
gamma1 = 0.5   # brightening
gamma2 = 1.0   # identity
gamma3 = 2.0   # darkening

s_gamma1 = r ** gamma1
s_gamma2 = r ** gamma2
s_gamma3 = r ** gamma3

# Plotting
plt.figure(figsize=(8,6))

# Log curve
plt.plot(r, s_log, label="Log Transform (s = c*log(r+1))", color="green")

# Gamma curves
plt.plot(r, s_gamma1, label="Gamma = 0.5 (Brighten)", color="blue")
plt.plot(r, s_gamma2, label="Gamma = 1 (Identity)", color="black", linestyle="--")
plt.plot(r, s_gamma3, label="Gamma = 2 (Darken)", color="red")

# Formatting
plt.title("Log vs Power Law (Gamma) Transformations")
plt.xlabel("Input Intensity (r)")
plt.ylabel("Output Intensity (s)")
plt.legend()
plt.grid(True)
plt.show()
```
